### Filter Data
Filter the `train_df` data to your selected store and products. Create a date range from 2016-01-01 to 2017-08-15 for consistency.

````python
# TODO: Filter train_df for selected store and date range 2016-01-01 to 2017-08-15

# Convert date column to datetime if needed
if train_df is not None:
    train_df['date'] = pd.to_datetime(train_df['date'])

# Filter for date range
start_date = '2016-01-01'
end_date = '2017-08-15'

# TODO: Create filtered datasets for each product family
product1_data = train_df[
    (train_df["store_nbr"] == selected_store) &
    (train_df["family"] == product_family_1) &
    (train_df["date"] >= start_date) &
    (train_df["date"] <= end_date)
].copy()

print("\n" + "="*80)
print("CHECKPOINT: Data filtered successfully")
print(f"Product data shape: {product1_data.shape if product1_data is not None else 'Not filtered'}")
print("="*80)
````

### Aggregate Daily Sales
Aggregate daily sales and handle missing dates.

````python
# Group by date and sum sales
daily_sales_1 = (
    product1_data.groupby("date", as_index=False)["sales"]
    .sum()
    .rename(columns={"sales": "total_sales"})
)

# Create complete date range
all_dates = pd.date_range(start=start_date, end=end_date, freq="D")
complete_df_1 = pd.DataFrame({"date": all_dates})

# Merge and fill missing days with 0
daily_sales_1 = complete_df_1.merge(daily_sales_1, on="date", how="left")
daily_sales_1["total_sales"] = daily_sales_1["total_sales"].fillna(0)

print("Daily sales shape:", daily_sales_1.shape)
print("Missing dates filled:", daily_sales_1["total_sales"].isna().sum())
daily_sales_1.head()
````

Analyze the seasonal components of your sales data.

### Day-of-Week Analysis

Add day of week to your data and calculate the average sales by day. Create a bar plot to visualize the weekday patterns.

````python
# Day-of-week analysis for Product 1
product1_dow = product1_ts.to_frame(name="total_sales").copy()
product1_dow["day_of_week"] = product1_dow.index.day_name()

dow_avg = (
    product1_dow.groupby("day_of_week")["total_sales"]
    .mean()
    .reindex(["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"])
)

plt.figure(figsize=(10,4))
plt.bar(dow_avg.index, dow_avg.values)
plt.title(f"Average Sales by Day of Week - Store {selected_store} - {product_family_1}")
plt.xlabel("Day of Week")
plt.ylabel("Average Total Sales")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

dow_avg
````

### Monthly Seasonality Analysis

Calculate the average sales by month for both products. Once calculated, create a line plot showing monthly patterns for the product.

````python
# Monthly seasonality for Product 1
product1_monthly = product1_ts.to_frame(name="total_sales").copy()
product1_monthly["month"] = product1_monthly.index.month

month_avg = product1_monthly.groupby("month")["total_sales"].mean()

plt.figure(figsize=(10,4))
plt.plot(month_avg.index, month_avg.values, marker="o")
plt.title(f"Average Sales by Month - Store {selected_store} - {product_family_1}")
plt.xlabel("Month (1=Jan ... 12=Dec)")
plt.ylabel("Average Total Sales")
plt.xticks(range(1,13))
plt.tight_layout()
plt.show()

month_avg
````

### Holiday Impact Analysis

Compare the average sales on holidays compared to regular days.

````python
# Holiday impact: compare holiday vs non-holiday days
daily_df = product1_ts.reset_index()
daily_df.columns = ["date", "total_sales"]
daily_df["date"] = pd.to_datetime(daily_df["date"])

hol = holidays_df.copy()
hol["date"] = pd.to_datetime(hol["date"])

# mark days that have any holiday/event
holiday_dates = hol["date"].drop_duplicates()
daily_df["is_holiday"] = daily_df["date"].isin(holiday_dates)

impact = daily_df.groupby("is_holiday")["total_sales"].mean()
impact.index = ["Non-Holiday", "Holiday/Event"]

plt.figure(figsize=(6,4))
plt.bar(impact.index, impact.values)
plt.title(f"Holiday vs Non-Holiday Avg Sales - Store {selected_store} - {product_family_1}")
plt.ylabel("Average Total Sales")
plt.tight_layout()
plt.show()

impact
````

## Step 5: Build Simple Forecasts and Compare Methods

### Time Series Decomposition

````python
from statsmodels.tsa.seasonal import seasonal_decompose

# Weekly seasonality (period=7) makes sense for daily grocery sales
decomposition1 = seasonal_decompose(product1_ts, model="additive", period=7)
decomposition1
````

````python
fig = decomposition1.plot()
plt.tight_layout()
plt.show()
````

### Build Forecasts
Create a baseline forecast using the average of the last 30 days and a season naive forcast by repeating the last 7 days' pattern.

````python
test_days = 30
train = product1_ts.iloc[:-test_days]
test = product1_ts.iloc[-test_days:]

# Method 1: Baseline forecast (30-day average)
baseline_value = train.iloc[-30:].mean()
baseline_forecast = np.repeat(baseline_value, test_days)

# Method 2: Seasonal Naive (repeat last 7 days pattern)
last_week = train.iloc[-7:].values
seasonal_naive_forecast = np.tile(last_week, int(np.ceil(test_days / 7)))[:test_days]

# Put forecasts into a DataFrame aligned to test dates
forecast_df = pd.DataFrame({
    "actual": test.values,
    "baseline_30day_avg": baseline_forecast,
    "seasonal_naive": seasonal_naive_forecast
}, index=test.index)

forecast_df.head()
````

### Calculate Forecast Accuracy

````python
# Calculate RMSE for both methods
# TODO: Calculate RMSE for baseline and seasonal forecasts

# Product 1
baseline_rmse = np.sqrt(mean_squared_error(test.values, baseline_forecast))
seasonal_rmse = np.sqrt(mean_squared_error(test.values, seasonal_naive_forecast))
````

````python
# Create comparison table
comparison_df = pd.DataFrame({
    "Method": ["30-Day Average", "Seasonal Naive (7-day repeat)"],
    "RMSE": [baseline_rmse, seasonal_rmse]
}).sort_values("RMSE")

# % improvement: how much better the best method is vs the baseline
best_rmse = comparison_df["RMSE"].min()
improvement_vs_baseline = (baseline_rmse - best_rmse) / baseline_rmse * 100

print("\n" + "="*60)
print("FORECAST PERFORMANCE COMPARISON")
print("="*60)
display(comparison_df)
print(f"\nBest method improves vs baseline by: {improvement_vs_baseline:.2f}%")
print("="*60)
````

### Visualize Forecasts

````python
# Visualize forecasts (last 60 days of train + 30 days test)
history_days = 60

history = train.iloc[-history_days:]   # last 60 days of training
test_series = test                     # last 30 days (actual)

plt.figure(figsize=(14, 6))

# Plot history + actual test
plt.plot(history.index, history.values, label="Train (last 60 days)")
plt.plot(test_series.index, test_series.values, label="Actual (test)")

# Plot forecasts aligned to the test dates
plt.plot(test_series.index, baseline_forecast, label="30-Day Average Forecast")
plt.plot(test_series.index, seasonal_naive_forecast, label="Seasonal Naive Forecast")

plt.title(f"Forecast Comparison - Store {selected_store} - {product_family_1}")
plt.xlabel("Date")
plt.ylabel("Total Sales")
plt.legend()
plt.tight_layout()
plt.show()
````
